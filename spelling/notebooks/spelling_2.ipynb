```python
def word_to_phonemes(word):
    """
    Convert a word to its ARPAbet phonemes and analyze its structure.
    """
    phoneme_list = pronouncing.phones_for_word(word)

    if not phoneme_list:
        return f"Word '{word}' not found in CMUdict."

    # Take the first pronunciation (CMUdict sometimes has multiple)
    phonemes = phoneme_list[0]
    
    # Split phonemes into a list
    phoneme_tokens = phonemes.split()
    
    # Count vowels and consonants
    vowels = {"IY", "IH", "EY", "EH", "AE", "AA", "AO", "OW", "UH", "UW", "ER", "AX", "AH", "AY", "AW", "OY"}
    vowel_count = sum(1 for phoneme in phoneme_tokens if phoneme[:2] in vowels)
    consonant_count = len(phoneme_tokens) - vowel_count

    # Count syllables (stress markers 0, 1, 2 indicate syllables)
    syllable_count = sum(1 for phoneme in phoneme_tokens if phoneme[-1].isdigit())

    # Display results
    result = {
        "Word": word,
        "Phonemes": phonemes,
        "Vowel Count": vowel_count,
        "Consonant Count": consonant_count,
        "Syllable Count": syllable_count
    }

    return result

# Example Usage
word = "hello"
result = word_to_phonemes(word)
print(result)
```

```python
import requests
import wikipediaapi
import json
import time

# Define a User-Agent string
HEADERS = {
    "User-Agent": "MyWikipediaScraper/1.0 (your_email@example.com)"
}

def get_trending_wikipedia_articles(limit=10):
    """
    Fetch trending Wikipedia articles from the Wikipedia API.
    """

    url = "https://en.wikipedia.org/api/rest_v1/feed/featured/2025/02/22"  # Replace with today's date
    response = requests.get(url, headers=HEADERS)
    
    if response.status_code != 200:
        print("Failed to fetch trending articles.")
        return []
    
    data = response.json()
    articles = [article['title'] for article in data.get('mostread', {}).get('articles', [])]
    return articles[:limit]  # Return top N articles

def fetch_wikipedia_text(title):
    """
    Fetch the full text of a Wikipedia article.
    """

    wiki = wikipediaapi.Wikipedia('en', headers=HEADERS)
    page = wiki.page(title)
    
    if not page.exists():
        return None
    
    return page.text

def save_dataset(articles_data, filename="data/wikipedia_dataset.json"):
    """
    Save collected Wikipedia articles as a JSON file.
    """

    with open(filename, "w", encoding="utf-8") as f:
        json.dump(articles_data, f, indent=4)
    
    print(f"Dataset saved as {filename}")

# Main Execution
popular_articles = get_trending_wikipedia_articles(limit=500)  # Get top 500 articles
dataset = {}

for title in popular_articles:
    print(f"Fetching article: {title}...")
    time.sleep(1.0)
    text = fetch_wikipedia_text(title)
    if text:
        dataset[title] = text

# Save to JSON
save_dataset(dataset)
```

```python
def split_words(text):
    valid = "abcdefghijklmnopqrstuvwxyz'"
    valid += valid.upper()
    boundry = "()/,:;.?!*$- "

    current_word = ""
    all_words = []
    for char in text:
        if char in valid:
            current_word += char
        elif char in boundry and current_word != "":
            all_words.append(current_word)
            current_word = ""
        else:
            current_word = ""
    
    return all_words

for title, article in dataset.items():
    print(title)
    words = split_words(article)
    phonemes = [word_to_phonemes(word) for word in words]
    display(phonemes)
    break
```

```python
def is_vowel(phoneme):
    return phoneme[-1].isdigit()

def vowel_count(phonemes):
    return len([p for p in phonemes if is_vowel(p)])

def contains_vowel(phonemes):
    return vowel_count(phonemes) >= 1

def get_syllables(word):
    phonemes = pronouncing.phones_for_word(word)
    if phonemes is None or len(phonemes) == 0:
        raise ValueError(f"unknown word {word}")
    
    phonemes = phonemes[0].split()

    print(phonemes)
    # word is only a single sylable
    if vowel_count(phonemes) == 1:
        return [phonemes]

    syllable = []
    syllables = []
    for p in reversed(phonemes):
        if is_vowel(p) and contains_vowel(syllable):
            syllables.insert(0, syllable)
            syllable = []
        
        syllable.insert(0, p)
    
    if len(syllable) > 0:
        syllables.insert(0, syllable)

    return syllables


for title, article in dataset.items():
    print(title)
    words = split_words(article)
    for word in words:
        print("---")
        print(word)
        syllable = get_syllables(word)
        print(syllable)
    break
```

```python
from enum import Enum
import pronouncing

class Sonority(Enum):
    # Vowels (highest sonority)
    AA = 6
    AE = 6
    AH = 6
    AO = 6
    AW = 6
    AY = 6
    EH = 6
    ER = 6
    EY = 6
    IH = 6
    IY = 6
    OW = 6
    OY = 6
    UH = 6
    UW = 6

    # Glides (semi-vowels)
    W = 5
    Y = 5

    # Liquids
    L = 4
    R = 4

    # Nasals
    M = 3
    N = 3
    NG = 3

    # Fricatives
    DH = 2
    F = 2
    V = 2
    TH = 2
    S = 2
    Z = 2
    SH = 2
    ZH = 2
    HH = 2

    # Plosives (Stops)
    B = 1
    D = 1
    G = 1
    K = 1
    P = 1
    T = 1
    CH = 1
    JH = 1

def get_sonority(phoneme):
    """
    Returns the sonority score of an ARPAbet phoneme.
    """
    phoneme = phoneme.strip("012")  # Remove stress markers
    return Sonority[phoneme].value if phoneme in Sonority.__members__ else 0  # Default to 0 for unknown phonemes

def syllabify(phonemes):
    """
    Segments a list of phonemes into syllables using the Sonority Sequencing Principle (SSP).
    """
    syllables = []
    current_syllable = []

    for i, phoneme in enumerate(phonemes):
        sonority = get_sonority(phoneme)

        if i > 0:
            prev_sonority = get_sonority(phonemes[i - 1])
            
            # If the current phoneme has lower sonority than the previous, it marks a syllable boundary
            if sonority < prev_sonority:
                syllables.append(current_syllable)
                current_syllable = []

        current_syllable.append(phoneme)

    # Add the last syllable
    if current_syllable:
        syllables.append(current_syllable)

    return ["-".join(syl) for syl in syllables]

def get_word_syllables(word):
    """
    Converts a word into ARPAbet phonemes and applies syllabification.
    """
    phoneme_lists = pronouncing.phones_for_word(word.lower())

    if not phoneme_lists:
        return f"No pronunciation found for '{word}'"

    # Take the first pronunciation (CMU Dictionary can have multiple)
    phonemes = phoneme_lists[0].split()
    return syllabify(phonemes)

# Example Usage
words = ["syllable", "computer", "elephant", "phonetics", "beautiful"]

for word in words:
    print(f"{word}: {get_word_syllables(word)}")
```

```python
import pronouncing

def get_syllables(word):
    # check if there's an inner capital like in words McDonald or iPhone
    # this check dosen't catch single letter words but that's fine since
    # those words will just be a single syllable anyway
    if word[1:].lower() != word[1:]:
        chars = [" "+char if char.isupper() else char for char in word]
        words = "".join(chars).split()
        syllables = []
        for word in words:
            # this is important for words like D'Angelo
            word = word.removesuffix("'")
            syllables.extend(get_syllables(word))

        return syllables

    phonemens = pronouncing.phones_for_word(word.lower())
    if phonemens is None or len(phonemens) == 0:
        raise ValueError(f"unknown word '{word}'")

    phonemens= phonemens[0].split()
    sonorities = [get_sonority(p) for p in phonemens]

    # pre-process sonorities to handle special case for 'S'
    for i in range(len(phonemens)):
        p = phonemens[i]
        s = sonorities[i]
        s_before = sonorities[i - 1] if i > 0 else -1
        s_after = sonorities[i + 1] if i < len(phonemens) - 1 else -1

        # special case for english cause it's weird
        if p == "S" and s_before < s and s > s_after:
            sonorities[i] = -1

    syllable = []
    syllables = []
    for i in reversed(range(len(phonemens))):
        p = phonemens[i]
        s = sonorities[i]
        s_before = sonorities[i - 1] if i > 0 else s
        s_after = sonorities[i + 1] if i < len(phonemens) - 1 else s

        if is_vowel(p) and contains_vowel(syllable):
            syllables.insert(0, syllable)
            syllable= [p]
            continue

        syllable.insert(0, p)
        if s_before > s and s < s_after and contains_vowel(syllable):
            syllables.insert(0, syllable)
            syllable= []


    if len(syllable) > 0:
        syllables.insert(0, syllable)

    return syllables

# test cases
cases = [
    # long words
    ("championship", [['CH', 'AE1', 'M'], ['P', 'IY0'], ['AH0', 'N'], ['SH', 'IH2', 'P']]),

    # words with apostraphies
    ("tournament's", [['T', 'UH1', 'R'], ['N', 'AH0'], ['M', 'AH0', 'N', 'T', 'S']]),
    ("can't", [['K', 'AE1', 'N', 'T']]),
    ("won't", [['W', 'OW1', 'N', 'T']]),
    ("shouldn't", [['SH', 'UH1'], ['D', 'AH0', 'N', 'T']]),

    # words with inner capital letters
    ("McDonald", [['M', 'IH0', 'K'], ['D', 'AA1'], ['N', 'AH0', 'L', 'D']]),
    ("D'Angelo", [['D', 'IY1'], ['AE1', 'N'], ['JH', 'AH0'], ['L', 'OW2']]),
    ("iPhone", [['AY1'], ['F', 'OW1', 'N']]),
    ("eBay", [['IY1'], ['B', 'EY1']]),
    ("CoPilot", [['K', 'OW1'], ['P', 'AY1'], ['L', 'AH0', 'T']]),

    # words with adjacent vowels
    ("fire", [['F', 'AY1'], ['ER0']]),
    ("hour", [['AW1'], ['ER0']]),
    ("player", [['P', 'L', 'EY1'], ['ER0']]),

    # words with silent letters
    ("knight", [['N', 'AY1', 'T']]),
    ("aisle", [['AY1', 'L']]),
    ("subtle", [['S', 'AH1'], ['T', 'AH0', 'L']]),

    # words with schwa
    ("separate", [['S', 'EH1'], ['P', 'ER0'], ['EY2', 'T']]),
    ("camera", [['K', 'AE1'], ['M', 'ER0'], ['AH0']]),
    ("family", [['F', 'AE1'], ['M', 'AH0'], ['L', 'IY0']]),

    # words with consonant clusters
    ("strength", [['S', 'T', 'R', 'EH1', 'NG', 'K', 'TH']]),
    ("sprints", [['S', 'P', 'R', 'IH1', 'N', 'T', 'S']]),
    ("through", [['TH', 'R', 'UW1']]),
    ("blasts", [['B', 'L', 'AE1', 'S', 'T', 'S']]),

    # compound words
    ("football", [['F', 'UH1', 'T'], ['B', 'AO2', 'L']]),
    ("wallpaper", [['W', 'AO1', 'L'], ['P', 'EY2'], ['P', 'ER0']]),
    ("bookshelf", [['B', 'UH1', 'K'], ['SH', 'EH2', 'L', 'F']]),

    # unusual stress patterns
    ("guitar", [['G', 'IH0'], ['T', 'AA1', 'R']]),
    ("police", [['P', 'AH0'], ['L', 'IY1', 'S']]),

    # loan words
    ("croissant", [['K', 'W', 'AA2'], ['S', 'AA1', 'N', 'T']]),
    ("cliche", [['K', 'L', 'IY0'], ['SH', 'EY1']]),
    ("karaoke", [['K', 'EH2'], ['R', 'IY0'], ['OW1'], ['K', 'IY0']]),

    # exceptions to the sonority sequencing principal
    ("psalm", [['S', 'AA1', 'L', 'M']]),
    ("sphere", [['S', 'F', 'IH1', 'R']]),
    ("split", [['S', 'P', 'L', 'IH1', 'T']]),
    ("sprint", [['S', 'P', 'R', 'IH1', 'N', 'T']]),
    ("sphinx", [['S', 'F', 'IH1', 'NG', 'K', 'S']]),
    ("thrive", [['TH', 'R', 'AY1', 'V']]),
    ("free", [['F', 'R', 'IY1']]),
    ("twelfth", [['T', 'W', 'EH1', 'L', 'F', 'TH']]),
    ("lengths", [['L', 'EH1', 'NG', 'K', 'TH', 'S']]),
    ("strength", [['S', 'T', 'R', 'EH1', 'NG', 'K', 'TH']]),
]
print("running test cases")
fail = False
for case in cases:
    word = case[0]
    want = case[1]
    got = get_syllables(word)
    if got != want:
        print(f"\n{word}: incorrect syllables got {got} but wanted {want}")
        fail = True
    else:
        print(".", end="")

assert not fail, "one or more test cases failed"
```

```python
from pydantic import BaseModel

class Conn(BaseModel):
    left: str
    right: str
    count: int = 1

class Graph:
    def __init__(self):
        self.connections = []
    
    def add_connections(self, path):
        if len(path) == 1:
            return

        for conn in zip(path[:-1], path[1:]):
            self.__add_connection(conn[0], conn[1])

    def __add_connection(self, left, right):
        for conn in self.connections:
            if conn.left == left and conn.right == right:
                conn.count += 1
                return
        
        conn = Conn(left = left, right= right)
        self.connections.append(conn)

    def get_next(self, left):
        next = []
        for conn in self.connections:
            if conn.left == left:
                next.append(conn)
        
        return next
    
    def get_prev(self, right):
        prev = []
        for conn in self.connections:
            if conn.right == right:
                prev.append(conn)
        
        return prev
    
    def save(self, file_path):
        """
        Saves the graph to a file in newline-delimited json (nd_json/jsonl) file.

        Args:
            file_path (str): The path to the file where the graph will be saved.
        """

        try:
            with open(file_path, "w", encoding="utf-8") as f:
                for conn in self.connections:
                    json_conn = conn.model_dump_json()
                    f.write(json_conn + "\n")

        except (IOError, OSError) as e:
            print(f"Error saving graph to {file_path}: {e}")

    def load(self, file_path):
        """
        Loads the graph from an nd_json/jsonl file.

        Args:
            file_path (str): The path to the file to load graph from.
        """

        try:
            self.connections = [] # clear current connections before loading.
            with open(file_path, "r", encoding="utf-8") as f:
                for line in f:
                    if line.strip(): #check if the line is not empty
                        try:
                            conn = Conn.model_validate_json(line)
                            self.connections.append(conn)
                        except Exception as e:
                            print(f"Filed to create connection from json: {line.strip()}. Error: {e}")

        except FileNotFoundError:
            print(f"File not found: {file_path}")
        except (IOError, OSError) as e:
            print(f"Error loading connections from {file_path}: {e}")


graph = Graph()
graph.add_connections(["a", "b", "c", "d"])
graph.add_connections(["d", "a", "b", "a"])
graph.add_connections(["d", "a", "f", "c"])

print(graph.connections)
print(graph.get_next("a"))
print(graph.get_prev("c"))
```

```python
import polars as pl
import json

from tqdm import tqdm

def load_datset(filename="data/wikipedia_dataset.json"):
    """
    Load collected Wikipedia articles into a Python dict.
    """

    dataset = {}
    with open(filename, "r", encoding="utf-8") as f:
        raw_file = f.read()
        dataset = json.loads(raw_file)
    
    return dataset

dataset = load_datset()

syllable_graph = Graph()
syllable_data = []
unknown_words = []
for title, article in tqdm(dataset.items(), desc="adding syllables to graph", total=len(dataset)):
    words = split_words(article)
    for word in words:
        try:
            syllables = get_syllables(word)
        except Exception as e:
            if str(e).startswith("unknown word"):
                unknown_words.append(word)
            else:
                print(f"failed to generate syllables for word {word}: {e}")
            continue
        
        for syllable in syllables:
            # remove vowel priority
            syllable = [s[:2] for s in syllable]

            syllable.insert(0, "<START>")
            syllable.append("<END>")

            syllable_graph.add_connections(syllable)
            syllable_data.append({"article": title, "word": word, "syllables": syllables})

print("building syllable dataframe...")
syllable_df = pl.DataFrame(syllable_data)
syllable_df.write_ndjson("data/syllable_data.jsonl")
display(syllable_df)

print("saving syllable graph...")
print("syllable graph size", len(syllable_graph.connections))
syllable_graph.connections[:10]
syllable_graph.save("data/syllable_graph.jsonl")
```
